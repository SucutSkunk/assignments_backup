### 문제 1-1. 내적으로 고객별 가중합 계산하기

#### 문제 설명

교안의 고객 활동 데이터와 가중치입니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| A | 3 | 2 |
| B | 5 | 1 |
| C | 2 | 4 |
| 가중치 | 2 | 4 |

입력과 가중치의 순서는 **방문 횟수 → 장바구니 추가 횟수**입니다. 이번 문제는 편향을 더하지 않고 내적만 계산합니다.

$$x\cdot w=x_1w_1+x_2w_2$$

- $x_1$, $x_2$: 고객의 첫 번째·두 번째 입력값
- $w_1$, $w_2$: 같은 위치의 특성에 적용할 가중치
- $\cdot$: 같은 위치끼리 곱한 값을 모두 더하는 내적 기호

#### 수식 읽는 순서

`x₁w₁`은 ‘첫 입력값 × 첫 가중치’입니다. 두 글자를 붙여 썼지만 곱셈을 뜻합니다. `x₂w₂`도 같은 방식입니다. `x·w`는 이 곱셈 결과들을 모두 더한 값입니다.

‘대응하는 원소’는 **같은 순서에 있는 숫자끼리**라는 뜻입니다. 방문 횟수에는 방문 가중치를, 장바구니 횟수에는 장바구니 가중치를 곱하세요.

````python
# 문제 1-1. 내적으로 고객별 가중합 계산하기
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 설명을 참고하세요.

w= np.array([2, 4])
x_a= np.array([3, 2])
x_b= np.array([5, 1])
x_c= np.array([2, 4])
print("w: ", w)
print("x_a: ", x_a)
print("x_b: ", x_b)
print("x_c: ", x_c)
````

````python
print("x_a.shape: ", x_a.shape)
print("w.shape: ", w.shape)
print("내적 가능하다.")
````

````python
print("x_a @ w의 값: ", x_a @ w )
print("둘 다 14로 동일하다.")
````

**수식 및 계산 과정**

- 공식: $x \cdot w = x_1w_1 + x_2w_2$, 가중치 $w = [2, 4]$ (방문 횟수 가중치 2, 장바구니 추가 횟수 가중치 4)

- `x_a_W` (고객 A, $x_A = [3, 2]$)
  $x_A \cdot w = 3 \times 2 + 2 \times 4 = 6 + 8 = 14$

- `x_b_W` (고객 B, $x_B = [5, 1]$)
  $x_B \cdot w = 5 \times 2 + 1 \times 4 = 10 + 4 = 14$

- `x_c_W` (고객 C, $x_C = [2, 4]$)
  $x_C \cdot w = 2 \times 2 + 4 \times 4 = 4 + 16 = 20$

- 손계산 결과(14, 14, 20)가 `np.dot()` 출력과 같음을 확인했다.

### 문제 1-2. 내적의 부호와 코사인 유사도 이해하기

#### 문제 설명

교안의 간단한 방향 예시를 사용합니다. 아래의 모든 벡터는 0벡터가 아닙니다.

| 구분 | 벡터 |
|---|---|
| 기준 벡터 `a` | [1, 0] |
| 같은 방향 `b_same` | [2, 0] |
| 수직 방향 `b_perp` | [0, 2] |
| 반대 방향 `b_opposite` | [−2, 0] |

추가로 교안의 `u = [3, 4]`, `v = [6, 8]`의 코사인 유사도를 계산합니다.

$$\mathrm{cosine\ similarity}=\frac{u\cdot v}{\|u\|_2\times\|v\|_2}$$

분자는 두 벡터의 내적, 분모는 두 벡터의 L2 Norm을 곱한 값입니다.

| 결과 | 방향 관계 |
|---|---|
| 코사인 유사도 1 | 같은 방향 |
| 코사인 유사도 0 | 서로 수직 |
| 코사인 유사도 −1 | 반대 방향 |

> 교안의 코사인 유사도 목록 중 ‘반대 방향: 1’은 ‘반대 방향: −1’의 부호 오타이므로 바로잡아 적용했습니다.

#### 먼저 이해하기

- 벡터를 원점에서 출발하는 화살표로 생각하면, ‘방향’은 화살표가 향하는 쪽입니다.
- `[1,0]`과 `[2,0]`은 같은 쪽, `[1,0]`과 `[0,2]`는 직각, `[1,0]`과 `[−2,0]`은 서로 반대쪽을 향합니다.
- 90°는 직각, 180°는 완전히 반대 방향입니다. 각도를 구하는 코드는 작성하지 않고 표와 내적의 부호를 연결합니다.
- **분자**는 분수의 위쪽, **분모**는 아래쪽입니다. 코사인 유사도 식은 ‘내적 ÷ (첫 길이 × 둘째 길이)’로 읽습니다.
- L2 길이는 이전 강의처럼 ‘각 원소 제곱 → 더하기 → 제곱근’으로 구합니다. 예를 들어 √25는 5×5=25이므로 5입니다.

1~2번은 방향별 내적, 3~4번은 u·v의 코사인 유사도입니다. 두 부분을 차례로 진행하세요.

#### 요구사항

#### 답변 작성란

**수식 및 계산 과정**

- 작성: 위의 셀들에 작성했다.

**Q1. X와 W의 Shape는 무엇이며 행렬곱이 가능한 이유는 무엇인가요?**

- X와 W의 shape는 둘 다 (2, 2)이다.
- 행렬곱이 가능한 이유는 **X의 열 수와 W의 행 수**가 둘 다 2로 같기 때문이다. 

**Q2. 고객 A의 두 점수는 각각 얼마인가요? 선택한 행·열과 계산 과정을 적으세요.**

- 고객 A의 두 점수는 [11, 7]이다.
- 구매 관심도 점수: X의 1행 [2, 3]과 W의 1열 [1, 3]을 내적 → $2 \times 1 + 3 \times 3 = 11$
- 고객 활동 점수: X의 1행 [2, 3]과 W의 2열 [2, 1]을 내적 → $2 \times 2 + 3 \times 1 = 7$


**Q3. 고객 B의 두 점수와 Y의 Shape는 무엇인가요? 결과의 행과 열이 무엇을 뜻하는지 설명하세요.**

- 고객 B의 두 점수는 [7, 9]이다. ($4 \times 1 + 1 \times 3 = 7$, $4 \times 2 + 1 \times 1 = 9$)
- Y의 Shape는 (2, 2)이다.
- Y의 행은 고객(1행 A, 2행 B)을, 열은 출력 점수의 종류(1열 구매 관심도, 2열 고객 활동)를 뜻한다. 즉 고객 수 × 출력 점수 종류 수이다.

**Q4. X @ W와 X * W의 첫 행 첫 열 값은 각각 얼마인가요? 왜 다른가요?**

- X @ W의 첫 행 첫 열 값은 11, X * W의 첫 행 첫 열 값은 2이다.
- X @ W의 첫 행 첫 열 값은 X의 첫 행과 W의 첫 열의 내적의 결과값이고
- X * W의 첫 행 첫 열 값은 X의 1행 1열 갑소가 W의 1행 1열 값을 곱한 것의 결과이기 때문이다.

**4.**

- 항상 가능한 것은 아니며 가능하더라도 같은 결과도 아니다.

- 일반적으로 값이 같기 어렵다.
